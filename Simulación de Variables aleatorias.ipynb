### Distribución Weibull

$We(α,1)$

Para $\quad x\geq0: F(x)=1-e^{-x^\alpha}$

Aplicando el método de la transformada inversa:

$U = 1-e^{-x^\alpha}$

Tenemos que:

$x = (-\ln(U))^{1/\alpha}$

#### Pseudocódigo:

Generar U~Unif(0,1)

x=(-ln(U))^(1/alpha)

Regresa x

#### Código:

```python
import random
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Asignamos parametro

alpha = 2

# Tamaño de nuestra muestra

n = 10000

# Lista donde se guardaremos los datos

muestra = []

# Generamos variables aleatorias

for i in range(n):

    # Variable uniforme

    U = random.random()

    # Transformada inversa

    X = (-np.log(U))**(1/alpha)

    # Guardamos dato

    muestra.append(X)

# Mostramos los primeros 10 valores (redondeados)

print("Primeros 10 valores:\n")

print([round(float(x),4) for x in muestra[:10]])
```

#### Histograma:

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 30,
    density = True)

plt.title("Histograma de Weibull")

plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.grid()

plt.show()
```

Observamos que la mayoría de valores ocurre cerca del origen y después la frecuencia va disminuyendo.

#### Distribución acumulada

```python
# La muestra se oredena de menor a mayor, se calcula la proporción acumulada 

plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50,
    cumulative = True)

plt.title("Distribución acumulada Weibull")

plt.xlabel("x")
plt.ylabel("Frecuencia acumulada")

plt.grid()

# Mostramos

plt.show()
```

Observamos que la frecuencia acumulada crece al igual crecen los valores de la variable aleatoria, asi aproximamos el comportamiento de la función de distribución acumulada $F(x)$

#### Media, varianza y desviación estándar 

```python
# Media

media = np.mean(muestra)

# Varianza

varianza = np.var(muestra)

# Desviación estándar

desviacion = np.std(muestra)

print(f"Media: {media:.4f}")

print(f"Varianza: {varianza:.4f}")

print(f"Desviación estándar: {desviacion:.4f}")
```

### Distribución Gumbel 

$Gu(0,1)$

$F(x)=e^{-e^{-x}}$

Aplicando el método de la transformada inversa:

$U=e^{-e^{-x}}$

Tenemos que:

$x=-\ln(-\ln(U))$

#### Pseudocódigo:

Generar U~Unif(0,1)

x=-ln(-ln(U))

Regresar x

#### Código:

```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
# Tamaño de nuestra muestra

n = 10000

# Lista para guardar los datos

muestra = []

# Para generar variables aleatorias

for i in range(n):

    # Variable uniforme

    U = random.random()

    # Transformada inversa

    X = -np.log(-np.log(U))

    # Guardamos dato

    muestra.append(X)

print("Primeros 10 valores:\n")

print([round(float(x),4) for x in muestra[:10]])
```

#### Histograma:

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50)

plt.title("Histograma de la distribución Gumbel")

plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.grid()

plt.show()
```

Observamos una mayor cantidad de datos en el centro y una disminución hacia extremos.

#### Distribución acumulada:

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50,
    cumulative = True)

plt.title("Distribución acumulada Gumbel")

plt.xlabel("x")
plt.ylabel("Frecuencia acumulada")

plt.grid()

plt.show()
```

Observamos que la frecuencia acumulada crece al igual crecen los valores de la variable aleatoria, asi aproximamos el comportamiento de la función de distribución acumulada $F(x)$

#### Media, varianza y desviacion estandar:

```python
# Media

media = np.mean(muestra)

# Varianza

varianza = np.var(muestra)

# Desviación estándar

desviacion = np.std(muestra)

print(f"Media: {media:.4f}")

print(f"Varianza: {varianza:.4f}")

print(f"Desviación estándar: {desviacion:.4f}")
```

### Distribución Cauchy

$F(x)=\frac{1}{\pi}\tan^{-1}(x)+\frac{1}{2}$

Aplicando el método de la transformada inversa:

$U=\frac{1}{\pi}\tan^{-1}(x)+\frac{1}{2}$

Tenemos que:

$x=\tan\left(\pi\left(U-\frac{1}{2}\right)\right)$

#### Pseudocódigo:

Generar U~Unif(0,1)

x= tan(pi*(U-0.5))

Regresar x

#### Código:

```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
# Tamaño de nuestra muestra

n = 10000

# Lista donde guardaremos los datos

muestra = []

# Para generar variables aleatorias

for i in range(n):

    # Variable uniforme

    U = random.random()

    # Transformada inversa

    X = np.tan(np.pi*(U - 0.5))

    # Guardamos dato

    muestra.append(X)

#Mostramos

print("Primeros 10 valores:\n")

print([round(float(x),4) for x in muestra[:10]])
```

#### Histograma:

```python
# Histograma

plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50)

plt.title("Histograma de la distribución Cauchy")

plt.xlabel("x")
plt.ylabel("Frecuencia")
 
plt.grid()

plt.show()
```

Observamos una gran cantidad de datos en el centro de la gráfica, y una menor cantidad de datos en los extremos.

#### Distribución acumulada 

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50,
    cumulative = True)

plt.title("Distribución acumulada Cauchy")

plt.xlabel("x")
plt.ylabel("Frecuencia acumulada")

plt.grid()

plt.show()
```

Observamos que la frecuencia acumulada crece al igual crecen los valores de la variable aleatoria, asi aproximamos el comportamiento de la función de distribución acumulada $F(x)$

#### Media, varianza y desviacion estandar 

```python
# Media

media = np.mean(muestra)

# Varianza

varianza = np.var(muestra)

# Desviación estándar

desviacion = np.std(muestra)

print(f"Media: {media:.4f}")

print(f"Varianza: {varianza:.4f}")

print(f"Desviación estándar: {desviacion:.4f}")
```

### Distribución logística 

$L(0,1)$

La función de distribución acumulada es:

$F(x)=\frac{1}{1+e^{-x}}$

Tenemos que:

$x=\ln\left(\frac{U}{1-U}\right)$

#### Pseudocódigo:

Generar U~Unif(0,1)

x=ln(U/(1-U))

Regresar x

#### Código:

```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
# Tamaño de nuestra muestra

n = 10000

# Lista donde guardamos los datos

muestra = []

# Para generar variables aleatorias

for i in range(n):

    # Variable uniforme

    U = random.random()

    # Transformada inversa

    X = np.log(U/(1-U))

    # Guardamos dato

    muestra.append(X)

print("Primeros 10 valores:\n")

print([round(float(x),4) for x in muestra[:10]])
```

#### Histograma:

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50)

plt.title("Histograma de la distribución Logística")

plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.grid()

plt.show()
```

Observamos gran cantidad de datos en el centro, con una disminución casi simétrica de cada lado

#### Distribución acumulada:

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50,
    cumulative = True)

plt.title("Distribución acumulada Logística")

plt.xlabel("x")
plt.ylabel("Frecuencia acumulada")

plt.grid()

plt.show()
```

Observamos que la frecuencia acumulada crece al igual crecen los valores de la variable aleatoria, asi aproximamos el comportamiento de la función de distribución acumulada $F(x)$

#### Media, varianza y desviacion estandar 

```python
# Media

media = np.mean(muestra)

# Varianza

varianza = np.var(muestra)

# Desviación estándar

desviacion = np.std(muestra)

print(f"Media: {media:.4f}")

print(f"Varianza: {varianza:.4f}")

print(f"Desviación estándar: {desviacion:.4f}")
```

### Distribución Pareto

$Par(α)$

Para $x \geq 1: F(x)=1-x^{-\alpha}$

Tenemos que:

$x=(1-U)^{-1/\alpha}$

#### Pseudocódigo:

Generar U~Unif(0,1)
x=U^(-1/alpha)
Regresar x

#### Código:

```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
# Parámetro de alpha

alpha = 3

# Tamaño de nuestra muestra

n = 10000

# Lista donde guardamos los datos

muestra = []

# Para generar variables aleatorias

for i in range(n):

    # Variable uniforme

    U = random.random()

    # Transformada inversa

    X = U**(-1/alpha)

    # Guardamos dato

    muestra.append(X)

print("Primeros 10 valores:\n")

print([round(float(x),4) for x in muestra[:10]])
```

#### Histograma:

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50)

# Limitamos eje x para verlo mejor

plt.xlim(1,10)

plt.title("Histograma de la distribución Pareto")

plt.xlabel("x")
plt.ylabel("Frecuencia")

plt.grid()

plt.show()
```

Observamos muchos datos cerca del valor mínimo y la disminución mientras aumentan los valores de la variable aleatoria 

#### Distribución acumulada 

```python
plt.figure(figsize=(10,5))

plt.hist(
    muestra,
    bins = 50,
    cumulative = True)

# Limitar eje x

plt.xlim(1,10)

plt.title("Distribución acumulada Pareto")

plt.xlabel("x")
plt.ylabel("Frecuencia acumulada")

plt.grid()

plt.show()
```

Observamos que la frecuencia acumulada crece al igual crecen los valores de la variable aleatoria, asi aproximamos el comportamiento de la función de distribución acumulada $F(x)$

#### Media, varianza y desviacion estandar 

```python
# Media

media = np.mean(muestra)

# Varianza

varianza = np.var(muestra)

# Desviación estándar

desviacion = np.std(muestra)

print(f"Media: {media:.4f}")

print(f"Varianza: {varianza:.4f}")

print(f"Desviación estándar: {desviacion:.4f}")
```
