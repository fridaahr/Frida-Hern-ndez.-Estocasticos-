```python
import numpy as np
import math
```

```python
# Definimos la matriz de tasas de transición R (ejercicio 1)
R = np.array([
    [0, 2, 3, 0],
    [4, 0, 2, 0],
    [0, 2, 0, 2],
    [1, 0, 3, 0]
], dtype=float)

# Definimos r (r = 6)
r = 6.0

print("Matriz de tasas R:")
print(R)
```

```python
# Construimos la matriz estocástica uniformizada P gorrito

# Calculamos las tasas totales de salida (suma de cada fila)
r_i = np.sum(R, axis=1)

N = R.shape[0]
P_gorr = np.zeros((N, N))

# Aplicamos la definición
for i in range(N):
    for j in range(N):
        if i == j:
            P_gorr[i, j] = 1.0 - (r_i[i] / r)
        else:
            P_gorr[i, j] = R[i, j] / r

print("Matriz estocástica P gorrito:")
print(P_gorr)
```

#### Ejercicio 3: Aproximación Numérica con Truncamiento Fijo ($M$)

Vamos a evaluar la matriz de probabilidad de transición $P(t)$

Para acotar:
$M = \max(\{ \lceil rt + 5\sqrt{rt} \rceil, 20 \})$

Parámetros de simulación calculados:
* Para $t = 0.5$ ($rt = 3$): $M = \max(\{12, 20\}) = 20$
* Para $t = 1.0$ ($rt = 6$): $M = \max(\{19, 20\}) = 20$
* Para $t = 5.0$ ($rt = 30$): $M = \max(\{58, 20\}) = 58$

```python
def aproximar_P_fijo(t, M_fijo):

    # Calculando P(t)
    rt = r * t
    N = P_gorr.shape[0]

    #Inivializamos B con ceros
    B = np.zeros((N, N))
    
    for k in range(M_fijo + 1):

        
        # Término de Poisson escalado
        # Usamos np.exp(-rt) y potencias directas, ya que M=20 es pequeñ
        coef_poisson = np.exp(-rt) * (rt**k) / math.factorial(k)

        # Potencia de la matriz P gorrito^k
        P_gorr_k = np.linalg.matrix_power(P_gorr, k)

        # Acumulamos en la matriz solución
        B += coef_poisson * P_gorr_k
        
    return B

# Resultados:

#Ejercicio 3.1
print("Ejercicio 3.1")

P_05 = aproximar_P_fijo(0.5, M_fijo=20)
print(f"\nMatriz P(0.5) con M=20:\n{P_05}")
P_1 = aproximar_P_fijo(1.0, M_fijo=20)
print(f"Matriz P(1.0) con M=20:\n{P_1}\n")
P_5 = aproximar_P_fijo(5.0, M_fijo=58)
print(f"Matriz P(5.0) con M=58:\n{P_5}\n")


# Ejercicio 3.2 (Verificación de Chapman-Kolmogorov)
print("Ejercicio 3.2")

# Multiplicamos P(0.5) por sí misma
P_05_cuadrado = P_05 @ P_05

print("P(0.5) * P(0.5):")
print(P_05_cuadrado)

# Calculamos la diferencia max absoluta entrada por entrada
diff_max = np.max(np.abs(P_1 - P_05_cuadrado))
print(f"\nDiferencia máxima absoluta: {diff_max:.2e}")
```

Al principio ($t=0.5$ y $t=1.0$), las probas cambian mucho dependiendo de dónde empiece la cadena. Pero a ($t=5.0$), todas las filas se vuelven iguales a $[0.2, 0.2, 0.4, 0.2]$, es decir ya se estabilizó y alcanzó su distribución estacionaria.

Multiplicar la matriz de $t=0.5$ por sí misma nos da la matriz de $t=1.0$. La diferencia es de apenas $5.82 \times 10^{-7}$, lo que demuestra el teorema de: $P(0.5) \times P(0.5) = P(1.0)$

#### Ejercicio 4: Algoritmo de uniformización eficiente por tolerancia

Aquí implementamos el pseudocódigo, el cual calcula dinámicamente el valor de $M$ óptimo deteniéndose cuando la masa de probabilidad acumulada supera $1 - \epsilon$. Esto garantiza que el error sea menor a la tolerancia $\epsilon$.

```python
def algoritmo_uniformizacion_tolerancia(t, epsilon=0.00001):
    
    # Pseucodigo:
    rt = r * t
    N = P_gorr.shape[0]
    
    # Inicialización de variables
    A = np.copy(P_gorr)          # Matriz para la siguiente potencia
    c = np.exp(-rt)             # Primer término de Poisson (k=0)
    B = np.exp(-rt) * np.eye(N) # Inicialización de la solución con c*I
    sum_prob = np.exp(-rt)      # Suma acumulada de probabilidades de Poisson
    k = 1                       # Contador de iteraciones
    
    # Ciclo iterativo guiado por la tolerancia del error
    while sum_prob < (1.0 - epsilon):
        
        # Propiedad recursiva del coeficiente de Poisson
        c = c * (rt / k)
        
        # Acumulamos el término actual en la matriz solución B
        B = B + c * A
        
        # Preparamos la matriz A para la siguiente potencia (A = A * P_gorr)
        A = A @ P_gorr
        
        # Acumulamos la masa de probabilidad
        sum_prob = sum_prob + c
        
        # Incrementamos el contador
        k += 1
        
    # El valor final de k al salir del ciclo corresponde al M dinámico alcanzado
    M_alcanzado = k - 1 
    return B, M_alcanzado

# Resultado
print("Ejercicio 4.2")
for t_val in [0.5, 1.0, 5.0]:
    matriz_res, m_optimo = algoritmo_uniformizacion_tolerancia(t_val, epsilon=0.00001)
    print(f"\nPara t = {t_val}: el algoritmo se detuvo en M = {m_optimo}")
    print(f"Matriz aproximada:\n{matriz_res}")
```

Observamos que con los dos métodos nos da el mismo resultado.

En lugar de adivinar un valor de $M$ fijo como en el ejercicio 3, while calcula de forma automática el número exacto de términos necesarios para que el error sea menor a $0.00001$.
